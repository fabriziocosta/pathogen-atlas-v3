# Phase 2a · Selected-domain PubMed pilot

Runs a small reproducible sample from the groups selected in `configs/default.yaml`. Pilot data are written to a separate cache and release so they cannot replace Phase 2 outputs. The report shows query totals, hits, incomplete searches, and abstract coverage.

```python
import os
import pandas as pd
from IPython.display import display
from pathogen_atlas_v3.abstracts.pubmed import resolve_ncbi_email, _ncbi_key_from_keychain
from pathogen_atlas_v3.common.config import selected_organism_groups
from pathogen_atlas_v3.common.io import load_yaml
from pathogen_atlas_v3.pipelines.phase2_pilot import run_phase2_pilot

CONFIG = load_yaml("configs/default.yaml")
GROUPS = selected_organism_groups(CONFIG)
print("Selected groups:", GROUPS)
print("NCBI contact available:", bool(resolve_ncbi_email()))
print("NCBI API key available:", bool(os.environ.get("NCBI_API_KEY") or _ncbi_key_from_keychain()))
PER_GROUP = 10
SEED = 42
MAX_FETCH_BATCHES = 8
RETMAX_PER_QUERY = 500
BATCH_SIZE = 100
MAX_WORKERS = 3
REQUEST_DELAY_SECONDS = 0.5
PILOT_ID = "pilot_selected_domains"
```

## Run the bounded pilot

This uses the same PubMed client and logging as Phase 2 while limiting each species search and the number of abstract-fetch batches. Each search/fetch log line includes an ETA.

```python
RUN_PILOT = False
if RUN_PILOT:
    PILOT_RESULT = run_phase2_pilot(
        per_group=PER_GROUP, seed=SEED, pilot_id=PILOT_ID,
        max_fetch_batches=MAX_FETCH_BATCHES, retmax=RETMAX_PER_QUERY,
        batch_size=BATCH_SIZE, max_workers=MAX_WORKERS,
        request_delay_seconds=REQUEST_DELAY_SECONDS, log=True,
    )
    display(PILOT_RESULT)
else:
    print("Pilot preflight only. Set RUN_PILOT=True to make bounded NCBI requests.")
```
